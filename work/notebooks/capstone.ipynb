## 1. Question
Can machine learning identify webpages that are likely to require a content refresh by using historical Search Intelligence data from the FlyRank warehouse?

This project supports the decision of helping SEO teams decide which webpages should be reviewed first. Instead of manually checking thousands of webpages, the model ranks pages based on their likelihood of needing a content refresh.

The project uses the FlyRank Search Intelligence warehouse dataset. Daily search performance data was aggregated into 30-day windows, and features such as impressions, clicks, average position, CTR, content age, and days since the last update were created. A Random Forest Classifier was trained to predict whether a webpage is likely to experience a significant decline in impressions. The results are intended to support SEO decisions and should be used as decision support rather than automatic decisions.

## 2. Data

This project uses the FlyRank ML Internship Warehouse dataset from Hugging Face.

The main table I am working with is fact_content_daily_performance.

It is a table with a lot of information.

The original warehouse has an amount of daily records. About 79 million.

I did not use all of these records for training.

Instead I used DuckDB to aggregate the data.

Now each row in my dataset represents one webpage over a 30-day time window.

I created some features from the aggregated data.

These features are:

* Total impressions

* clicks

* Average position

* Total sessions

* Total users

* Content age

* Days since last update

* Click Through Rate

I also created a target variable called needs_refresh.

I did this by comparing the impressions of a webpage over two 30-day periods.

If the impressions dropped by than 20% I labeled the webpage as 1, which means it Needs Refresh.

If the impressions did not drop much I labeled it as 0 which means No Refresh.

I want to make it clear that I did not use any information in this project.

This includes things, like client names, URLs and search queries.

## 3. Methodology

The objective of this project is to identify webpages that may need a content refresh.

First, DuckDB was used to read the warehouse data directly from Hugging Face and aggregate daily records into 30-day windows.

After creating the aggregated dataset, missing values were handled and the target variable was created using the 20% decline rule.

A simple baseline was created using fixed rules based on CTR, impressions, and content age.

Then a Random Forest Classifier was trained using the selected features. The model was evaluated using both a random train-test split and a grouped-by-client validation split.

The grouped validation was used to reduce data leakage because pages from the same client were kept together during testing.

## 4. Results (vs baseline)
The Random Forest model was compared with a simple rule-based baseline using the same dataset.

| Method | Description | Result |
|--------|-------------|--------|
| Baseline | Fixed rules based on impressions, CTR, and content age | Simple ranking of webpages |
| Random Forest Classifier | Machine learning model trained on aggregated Search Intelligence features | Accuracy = 80.45% |

The Random Forest model performed better than the baseline because it learned relationships between multiple SEO features instead of relying on fixed rules.

A grouped-by-client validation was also performed to evaluate the model on unseen clients. The grouped validation achieved approximately 80% accuracy, showing that the model can generalize reasonably well while providing a more realistic evaluation.

The results suggest that the model can help prioritize webpages for review. However, the recommendations should be used as decision support and should always be reviewed by SEO specialists before taking action.

## 5. Limitations

This project was made with Search Intelligence information from the FlyRank warehouse.

The model looks for patterns in the information we have. It does not tell us why the performance of a webpage changes. There are things that can affect this that are not in our information like when Google changes its rules or when things are different at certain times of the year or what our competitors are doing or changes in our business.

We should use the model to help us make decisions. It shows us which webpages might need some work. An SEO specialist should always make the final decision after looking at the webpage.

The results depend on the time period we choose. What information we have. The model might work better or worse when we get information so we should check it often and update it when we need to. The FlyRank warehouse and Search Intelligence data are important, for this project and the model.

## 6. Ranked recommendations

The trained Random Forest model predicts whether a webpage is likely to require a content refresh. Pages predicted as high risk should be reviewed first.

| Priority | Recommended Action | Reason Code |
|----------|--------------------|-------------|
| High | Refresh Immediately | HIGH_REFRESH_SCORE |
| Medium | Review Content | POSSIBLE_DECLINE |
| Medium | Monitor Closely | MODERATE_RISK |
| Low | Monitor | LOW_RISK |
| Lowest | No Action | STABLE_PAGE |

These recommendations help SEO teams focus on the most important webpages instead of manually checking thousands of pages. The recommendations should be used as guidance and not as automatic decisions.
